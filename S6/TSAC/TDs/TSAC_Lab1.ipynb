```r
library(astsa)
data(jj)
y <- log(jj)

v <- filter(y, rep(1/4,4),sides = 1)

plot(y, type = "l", lty = 1,
     main = "Johnson & Johnson Log Earnings with Moving Average Filter",
     xlab = "Time",
     ylab = "Log Earnings",
     col = "black")

# Overlay the filtered series as a dashed line
lines(v, lty = 2, col = "blue", lwd = 2)

# Add a legend
legend("topleft",
       legend = c("Original Y_t", "Filtered V_t (MA-4)"),
       lty = c(1, 2),
       col = c("black", "blue"),
       lwd = c(1, 2))
```
