### Exercise 2: Text Statistics

Write a program that tracks how often the following main characters appear together within the same paragraph: Alice, Queen, King, Rabbit, Hatter, Duchess.

Output: Display character pairs and the frequency of their co-occurrence.

```python
import nltk
import random
from nltk.tokenize import word_tokenize
from nltk import FreqDist, pos_tag
from collections import defaultdict, Counter
import matplotlib.pyplot as plt
from nltk.corpus import gutenberg, stopwords

# Load the "Alice in Wonderland" text from NLTK's Gutenberg corpus
alice_text = gutenberg.raw('carroll-alice.txt')
stop_words = set(stopwords.words('english'))

def analyze_character_cooccurrence():
    """Analyze character co-occurrence using NLTK sentence tokenization"""
    print("\n=== Exercise 2: Character Co-occurrence Analysis ===")
    paragraphs = alice_text.split('\n\n')
    print(f"Number of paragraphs: {len(paragraphs)}")

    main_characters = {'Alice', 'Queen', 'King', 'Rabbit', 'Hatter', 'Duchess'}
    counts = Counter()

    for p in paragraphs:
        words = [w.capitalize() for w in word_tokenize(p)]
        characters_in_sentence = set(words) & main_characters
        if len(characters_in_sentence) > 1:
            for char in characters_in_sentence:
                counts[char] += words.count(char)

    print(counts)

analyze_character_cooccurrence()
```

## Exercise 3: Next Word Prediction and Sentence Generation

1. **Next Word Prediction**:
   - Develop a program that predicts the **next word** in a text given a specific word.
   
   Hint: Track the words that most commonly appear **immediately after** the given word in the text.
   
2. **Randomized Word Selection**:

3. **Generate a Sentence**:
   - Starting with a given word, generate a sentence of 10 words by predicting the next word based on the previous word in the sentence. Randomly choose the next word from the top 5 most frequent followers.

4. **Optional Question: Next Word Prediction (N-grams)**:
  - Extend the program to predict the next word in a sequence, conditioning the prediction on the words immediately preceding it:
      - Trigram Prediction (2-word History): Modify the word prediction process to use a history of the two previous words (e.g., 'I was'), predict the next word (e.g., 'going'). The next word should be chosen randomly from the top 5 most frequent words that follow that specific two-word sequence in the text.
    - Four-gram Prediction (3-word History): Further modify the prediction process to use a history of the three previous words (e.g., 'I was going'), predict the next word (e.g., 'home'). The next word should be chosen randomly from the top 5 most frequent words that follow that specific three-word sequence in the text.

### Example Output:
- Given the word **"Alice"**, you might generate sentences like:
  - "Alice was beginning to feel very tired of sitting by the"



```python
import nltk
import random
from nltk.tokenize import word_tokenize
from nltk import FreqDist, pos_tag
from collections import defaultdict, Counter
import matplotlib.pyplot as plt
from nltk.corpus import gutenberg, stopwords

# Load the "Alice in Wonderland" text from NLTK's Gutenberg corpus
alice_text = gutenberg.raw('carroll-alice.txt')
# NOTE: The original code used 'text' later, replacing with 'alice_text' for consistency
text = alice_text
stop_words = set(stopwords.words('english'))

# Tokenize the text and keep all words, including stop words
def preprocess_text(text):
    # Lowercase and tokenize so 'alice' matches 'Alice' in the text
    tokens = word_tokenize(text.lower())
    return tokens

tokens = preprocess_text(text)

# Function to track the words that come immediately after a given word (Unigram/Bigram model)
def analyze_next_word(tokens):
    next_word_dict = defaultdict(list)
    for i in range(len(tokens) - 1):
        current_word = tokens[i]
        next_word = tokens[i + 1]
        next_word_dict[current_word].append(next_word)
    return next_word_dict

# Function to track the words that come immediately after a pair of words (Trigram model)
def analyze_next_word_trigram(tokens):
    next_word_trigram_dict = defaultdict(list)
    for i in range(len(tokens) - 2):
        current_pair = (tokens[i], tokens[i + 1])
        next_word = tokens[i + 2]
        next_word_trigram_dict[current_pair].append(next_word)
    return next_word_trigram_dict

# Function to track the words that come immediately after three words (Four-gram model)
def analyze_next_word_fourgram(tokens):
    next_word_fourgram_dict = defaultdict(list)
    for i in range(len(tokens) - 3):
        current_triplet = (tokens[i], tokens[i + 1], tokens[i + 2])
        next_word = tokens[i + 3]
        next_word_fourgram_dict[current_triplet].append(next_word)
    return next_word_fourgram_dict


# Sentence generation using predicted next words (Bigram, based on 1 previous word)
def generate_sentence_bigram(start_word, next_word_dict, sentence_length=15):
    sentence = [start_word]
    current_word = start_word

    # Generate the rest of the sentence by predicting the next word based on the *last* word
    for _ in range(sentence_length - 1):
        if current_word in next_word_dict:
            next_word = random.choice(next_word_dict[current_word])
            sentence.append(next_word)
            current_word = next_word
        else:
            break

    return ' '.join(sentence)

# Sentence generation using Trigram prediction (based on 2 previous words)
def generate_sentence_trigram(start_pair, next_word_trigram_dict, sentence_length=15):
    sentence = list(start_pair) # start_pair is a tuple/list of two words

    # Generate the rest of the sentence by predicting the next word based on the *last two* words
    for _ in range(sentence_length - 2):
        current_pair = (sentence[-2], sentence[-1])
        if current_pair in next_word_trigram_dict:
            next_word = random.choice(next_word_trigram_dict[current_pair])
            sentence.append(next_word)
        else:
            break

    return ' '.join(sentence[:sentence_length])

# Sentence generation using Four-gram prediction (based on 3 previous words)
def generate_sentence_fourgram(start_triplet, next_word_fourgram_dict, sentence_length=15):
    sentence = list(start_triplet) # start_triplet is a tuple/list of three words

    # Generate the rest of the sentence by predicting the next word based on the *last three* words
    for _ in range(sentence_length - 3):
        current_triplet = (sentence[-3], sentence[-2], sentence[-1])
        if current_triplet in next_word_fourgram_dict:
            next_word = random.choice(next_word_fourgram_dict[current_triplet])
            sentence.append(next_word)
        else:
            break

    return ' '.join(sentence[:sentence_length])

# Example usage
start_word = 'alice'
start_pair = ('alice', 'was')
start_triplet = ('alice', 'was', 'beginning')


# Step 1: Analyze next word (Bigram/1-word context)
next_word_dict = analyze_next_word(tokens)

# Step 2: Analyze next word (Trigram/2-word context)
next_word_trigram_dict = analyze_next_word_trigram(tokens)

# Step 3: Analyze next word (Four-gram/3-word context)
next_word_fourgram_dict = analyze_next_word_fourgram(tokens)


# Step 4: Generate a sentence using Bigram prediction (1-word context)
print("Generated Sentence (Bigram/1-word Context):")
print(generate_sentence_bigram(start_word, next_word_dict))

# Step 5: Generate a sentence using Trigram prediction (2-word context)
print("\nGenerated Sentence (Trigram/2-word Context):")
print(generate_sentence_trigram(start_pair, next_word_trigram_dict))

# Step 6: Generate a sentence using Four-gram prediction (3-word context)
print("\nGenerated Sentence (Four-gram/3-word Context):")
print(generate_sentence_fourgram(start_triplet, next_word_fourgram_dict))
```

## Exercise 4: Chapter Vocabulary Profiler

In this exercise, you will create a program that evaluates the vocabulary complexity of each chapter in the story. This analysis will provide insights into the linguistic richness of the chapters by calculating various metrics.

### Objectives

1. **Average Word Length**:
   - Calculate the average length of words in each chapter. This metric helps determine the complexity of the vocabulary used.

2. **Number of Unique Words**:
   - Count the number of unique words in each chapter. This indicates the diversity of the vocabulary.

3. **Frequency of Complex Words**:
   - Identify and count the frequency of complex words, defined as words with more than 5 or 6 letters. This metric highlights the use of advanced vocabulary.

4. **Merging Metrics**:
   - Merge the three metrics—average word length, number of unique words, and frequency of complex words—into a single vocabulary score using the following formula:

   Vocabulary Score = (0.4 * Average Word Length) + (0.4 * Number of Unique Words) + (0.2 * Number of Complex Words)

   Apply this formula to generate a combined score for each chapter, allowing for easier comparison between them.


   Apply this formula to generate a combined score for each chapter, allowing for easier comparison between them.

```python
import nltk
from nltk.tokenize import word_tokenize
import matplotlib.pyplot as plt

# Load the "Alice in Wonderland" text from NLTK's Gutenberg corpus
from nltk.corpus import gutenberg
alice_text = gutenberg.raw('carroll-alice.txt')

def analyze_chapter_vocabulary():
    """Analyze vocabulary complexity of each chapter and sort them by complexity"""
    print("\n=== Exercise 4: Chapter Vocabulary Analysis ===")
    
    # Note: Using 'CHAPTER' splits the book into its 12 chapters 
    # (swap back to alice_text.split('\n\n') if you prefer paragraphs)
    chapters = alice_text.split('CHAPTER')[1:]
    
    avg_word_lengths = []
    unique_word_counts = []
    complex_word_counts = []
    vocabulary_scores = []

    for i, chapter in enumerate(chapters):
        words = word_tokenize(chapter)

        # 1. Calculate the average length of words in each chapter
        avg_length = sum(len(word) for word in words) / len(words) if words else 0
        avg_word_lengths.append(avg_length)

        # 2. Calculate the number of unique words in each chapter
        unique_words = set(words)
        unique_count = len(unique_words)
        unique_word_counts.append(unique_count)

        # 3. Frequency of complex words (words with more than 6 letters)
        complex_words = [word for word in words if len(word) > 6]
        complex_count = len(complex_words)
        complex_word_counts.append(complex_count)

        # 4. Merging metrics into a single Vocabulary Score
        score = (0.4 * avg_length) + (0.4 * unique_count) + (0.2 * complex_count)
        vocabulary_scores.append((i + 1, score, avg_length, unique_count, complex_count))

    # Sort chapters by Vocabulary Score (highest to lowest)
    vocabulary_scores.sort(key=lambda x: x[1], reverse=True)

    # Print the sorted results
    for chap_num, score, avg_len, unique_c, complex_c in vocabulary_scores:
        print(f"Chapter {chap_num:2d} | Score: {score:6.2f} | Avg Len: {avg_len:.2f} | Unique: {unique_c:4d} | Complex: {complex_c:3d}")

# Call the function
analyze_chapter_vocabulary()
```

## Exercise 5: Character Action Tracker

In this exercise, you will develop a program to identify and track the actions (verbs) most commonly associated with the main characters in the story. This analysis will help you understand the behaviors and activities of the characters throughout the narrative.

### Main Characters

For this analysis, we will focus on the following main characters: **Alice, Queen, King, Rabbit, Hatter, Duchess**.

### Objectives

1. **Identify Actions**:
   - Analyze the text to extract **verbs** associated with each main character. This involves tracking the occurrences of verbs that appear within a short context around the character's name.
   - **Hint**: Focus on identifying verbs that occur **within the next 4 words** after the character's name appears in a sentence.

2. **Count and Rank**:
   - Count how many times each verb is associated with each character and rank them to find the most frequent actions.

3. **Output Results**:
   - For each main character, print a list of the top verbs associated with them, along with their frequencies.


```python
import nltk
import random
from nltk.tokenize import word_tokenize
from nltk import FreqDist, pos_tag
from collections import defaultdict, Counter
import matplotlib.pyplot as plt
from nltk.corpus import gutenberg, stopwords

# Load the "Alice in Wonderland" text from NLTK's Gutenberg corpus
alice_text = gutenberg.raw('carroll-alice.txt')
stop_words = set(stopwords.words('english'))

def track_character_actions():
    """Track character actions using POS tagging"""
    print("\n=== Exercise 5: Character Action Tracker ===")

    main_characters = {'Alice', 'Queen', 'King', 'Rabbit', 'Hatter', 'Duchess'}
    character_verbs = defaultdict(Counter)

    # 1. Tokenize and POS-tag the text
    words = word_tokenize(alice_text)
    tagged_words = pos_tag(words)

    # 2. Find verbs within the next 4 words after a character's name
    for i, (word, tag) in enumerate(tagged_words):
        if word in main_characters:
            # Look at the next 4 words
            for next_word, next_tag in tagged_words[i + 1 : i + 5]:
                # In NLTK, all verb tags start with 'VB' (VB, VBD, VBG, VBN, VBP, VBZ)
                if next_tag.startswith('VB'):
                    character_verbs[word][next_word.lower()] += 1

    # 3. Print the top 5 verbs for each character
    for char in main_characters:
        print(f"\nTop actions for {char}:")
        for verb, count in character_verbs[char].most_common(5):
            print(f"  {verb}: {count}")

track_character_actions()
```

## Exercise 6: Sentence Type Classification

In this exercise, we will categorize sentences in the text based on their type, focusing on **Questions**, **Exclamations**, **Statements**, and **Imperatives**. By analyzing the structure of each sentence, we can gain a better understanding of how characters express themselves and how the narrative develops.

### Tasks

1. **Questions**:
   - Identify sentences that ask for information or clarification.
   - Specifically, we will focus on questions that start with one of the following words:
     - "Who"
     - "What"
     - "Why"
     - "Where"
   - Ensure that these sentences end with a question mark (`?`).

2. **Exclamations**:
   - Identify sentences that express strong emotions or surprise.
   - These sentences typically end with an exclamation mark (`!`).

3. **Statements**:
   - Identify sentences that declare facts or provide information.
   - These sentences usually start with a **noun** or **pronoun** (e.g., "Alice," "She," "The cat").
   - They end with a period (`.`).

4. **Imperatives**:
   - Identify sentences that give commands, requests, or instructions.
   - These sentences usually start with a **verb in its base form** (e.g., "Go," "Run," "Take").
   - They can end with a period (`.`) or an exclamation mark (`!`).


### Tasks to Complete:
- Extract and classify sentences from the text into the four categories mentioned above.
- Print the top 5 sentences from each category to understand the different sentence structures and their usage in the story.


```python
import nltk
import matplotlib.pyplot as plt
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk import pos_tag  # Added import for pos_tag


# Load the "Alice in Wonderland" text from NLTK's Gutenberg corpus
from nltk.corpus import gutenberg
alice_text = gutenberg.raw('carroll-alice.txt')

# Function to check if a word is a noun or pronoun
def is_noun_or_pronoun(word):
    """Check if a word is a noun or pronoun based on its part of speech tag."""
    return pos_tag([word])[0][1] in ['NN', 'NNS', 'NNP', 'NNPS', 'PRP', 'PRP$']


# Function to extract and categorize sentences
def categorize_sentences(text):
    """Categorize sentences into Questions, Exclamations, Statements, and Imperatives."""
    sentences = sent_tokenize(text)

    # Initialize categories
    categories = {'Questions': [], 'Exclamations': [], 'Statements': [], 'Imperatives': []}

    question_words = {'who', 'what', 'when', 'where', 'why', 'how'}

    for sentence in sentences:
        stripped_sentence = sentence.strip()
        if stripped_sentence.endswith('?'):
            categories['Questions'].append(stripped_sentence)
        elif stripped_sentence.endswith('!'):
            categories['Exclamations'].append(stripped_sentence)
        elif stripped_sentence.split()[0].lower() in question_words:
            categories['Questions'].append(stripped_sentence)
        elif is_noun_or_pronoun(stripped_sentence.split()[0]):
            categories['Statements'].append(stripped_sentence)
        else:
            categories['Imperatives'].append(stripped_sentence)

    return categories  # Added return statement

# Categorize the sentences in the text
categories = categorize_sentences(alice_text)

# Print the top 5 sentences from each category
for category, sents in categories.items():
    print(f"\n=== {category} (Total: {len(sents)}) ===")
    for sent in sents[:5]:
        print(f"  - {sent}")

# Function to plot the frequencies of the categories
def plot_category_frequencies(categories):
    """Plot the frequencies of different sentence categories."""
    category_counts = {category: len(sentences) for category, sentences in categories.items()}

    plt.figure(figsize=(10, 6))
    plt.bar(category_counts.keys(), category_counts.values(), color=['blue', 'orange', 'green', 'red'])
    plt.title('Frequencies of Sentence Categories in Alice in Wonderland')
    plt.xlabel('Sentence Categories')
    plt.ylabel('Number of Sentences')
    plt.show()


# Plot the frequencies
plot_category_frequencies(categories)
```

### Exercise 7: Analysis of Violent Actions and Death Mentions

In this exercise, you will develop a program to identify and analyze sentences in the story that contain mentions of violent actions or death. This analysis aims to explore the themes of violence and mortality within the narrative.

### Tasks

1. **Identify Relevant Sentences**:
   - Write a program that scans the text for sentences containing specific keywords related to violent actions or death. Use the following keywords as indicators:
     - **Death-Related Words**:
       - "kill"
       - "murder"
       - "die"
       - "death"
     - **Violent Action Words**:
       - "attack"
       - "stab"
       - "shoot"
       - "injure"
       - "hurt"
       - "blood"
       - "assault"

2. **Categorize Instances**:
   - After identifying sentences containing the relevant keywords, categorize them into two distinct groups:
     - **Death Indicators**: Sentences containing any of the death-related words.
     - **Violent Actions**: Sentences containing any of the violent action words.

3. **Display Results**:
   - For each category, print the top 10 sentences that correspond to each keyword. This will provide insights into how often these themes appear in the text and in what context.


```python
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import gutenberg

# Load Alice in Wonderland text from the NLTK Gutenberg corpus
alice_text = gutenberg.raw('carroll-alice.txt')

# Keywords related to death and violence
death_related_words = {"kill", "murder", "die", "death"}
violent_action_words = {"attack", "stab", "shoot", "injure", "hurt", "blood", "assault"}

def analyze_violence_and_death(text):
    """Identify and categorize sentences containing death and violent action keywords."""
    print("\n=== Exercise 7: Analysis of Violent Actions and Death Mentions ===")
    sentences = sent_tokenize(text)

    death_indicators = []
    violent_actions = []

    for sentence in sentences:
        # Tokenize and lowercase the words in the sentence
        words = set(word_tokenize(sentence.lower()))

        # Check if any death-related words appear in the sentence
        if words & death_related_words:
            death_indicators.append(sentence.strip())

        # Check if any violent action words appear in the sentence
        if words & violent_action_words:
            violent_actions.append(sentence.strip())

    # Display top 10 sentences for Death Indicators
    print(f"\n=== Death Indicators (Total: {len(death_indicators)}) ===")
    for sent in death_indicators[:10]:
        print(f"  - {sent}\n")

    # Display top 10 sentences for Violent Actions
    print(f"\n=== Violent Actions (Total: {len(violent_actions)}) ===")
    for sent in violent_actions[:10]:
        print(f"  - {sent}\n")

analyze_violence_and_death(alice_text)
```
