```python
import random

def simulate_events(n=6):
    count_A_B = 0
    count_A_notB = 0
    count_notA_B = 0
    count_notA_notB = 0

    for _ in range(n):
        uA = random.random()
        uB = random.random()

        A = uA < 0.3
        B = uB < 0.55

        if A and B:
            count_A_B += 1
        elif A and not B:
            count_A_notB += 1
        elif (not A) and B:
            count_notA_B += 1
        else:
            count_notA_notB += 1

    return (count_A_B, count_A_notB, count_notA_B, count_notA_notB)

results = simulate_events()
print("A ∩ B:", results[0])
print("A ∩ B̄:", results[1])
print("Ā ∩ B:", results[2])
print("Ā ∩ B̄:", results[3])
```
