```python
def generate(a,b,m,n,seed):
    random_numbers = []
    current_value = seed

    for _ in range(n):
        current_value = (a * current_value + b) % m
        random_numbers.append(current_value / m)
    return random_numbers
```

```python
import matplotlib.pyplot as plt

def lcg_period(a, b, m, seed):
    seen = set()
    y = seed
    count = 0
    
    while y not in seen:
        seen.add(y)
        y = (a * y + b) % m
        count += 1
    
    return count

a = 17
b = 43
m = 100
n = 30

for seed in range(1, 28):
    print(f"Seed {seed} → Period = {lcg_period(a, b, m, seed)}")


fig, axes = plt.subplots(7, 4, figsize=(12, 16))
axes = axes.flatten()

for seed in range(28):
    nums = generate(a, b, m, n, seed)
    ax = axes[seed]
    ax.hist(nums, bins=10, edgecolor='black')
    ax.set_title(f"Seed {seed}", fontsize=8)
    ax.tick_params(axis='both', which='major', labelsize=6)

plt.tight_layout()
plt.show()
```
